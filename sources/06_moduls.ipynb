Importing a sub model datetime from parent module datetime 

```python
from datetime import datetime
```

```python
#prints the date and time right now
print(datetime.now())
```

```python
#prints just the current date not time 
print(datetime.now().date())
```

```python
#prints current time only
print(datetime.now().time())
```

```python
#prints year only
print(datetime.now().year)
```

```python
#prints month only
print(datetime.now().month)
```

```python
#prints day only
print(datetime.now().day)
```

```python
#string format for time. changing the format of the output in strfttime(format you want)
print(datetime.now().strftime('%d/%m/%y'))
```

```python
#show day of the week
#%b means month name
#%m means month number 
#%y means 2 digits for year
#%Y means 4 digits for year
print(datetime.now().strftime('%A %d, %b, %Y'))
```
