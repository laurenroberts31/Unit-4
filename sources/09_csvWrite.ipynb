
Writing csv files 

```python
import csv
```

```python
makes = ['Toyota', 'Ford', 'BMW']
models =['Rav4', 'F-150', 'X6']
max_speed = [220, 200, 260]
powers = [195, 400, 360]
```

```python
#combinging lists together to make one dataset 
cars_list = list(zip(makes, models, max_speed, powers))
```

```python
#exporting dataset to csv file
#creating the file cars_data.csv in the datasets folder
#specify 'w' --> means write mode, newline='' --> to avoid blank lines between rows
#writer.writerows() --> to write multiple rows at once
#writer.writerow() --> to write a single row
with open('../datasets/cars_data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    #writing the header row
    writer.writerow(['Make', 'Model', 'Max Speed', 'Power'])
    #writing the data rows
    writer.writerows(cars_list)
```
